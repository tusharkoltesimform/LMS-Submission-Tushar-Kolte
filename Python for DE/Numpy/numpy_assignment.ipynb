### Task 0: Logging Setup
Configure application-level logging with the following requirements:

*   Log the start and completion of the overall pipeline.
*   Log the start and end of each task.
*   Capture warnings, errors, and exceptional conditions.

```python
import logging
import pandas as pd
import numpy as np

logger = logging.getLogger('my_logger')
logger.setLevel(logging.DEBUG)

if logger.hasHandlers():
    logger.handlers.clear()

filehandler = logging.FileHandler('my_logger.log')
filehandler.setLevel(logging.DEBUG)

consolehandler = logging.StreamHandler()
consolehandler.setLevel(logging.DEBUG)

formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')

filehandler.setFormatter(formatter)
consolehandler.setFormatter(formatter)

logger.addHandler(filehandler)
logger.addHandler(consolehandler)

logger.info('Started the numpy assignment')
```

### Task 2: Data Cleaning
*   Identify missing values across all columns.
*   Replace missing `Age` values using the median of the available data.
*   Replace missing `Salary` values using the mean of the available data.
*   Detect and remove duplicate employee records based on `Employee_ID`.
*   Standardize column naming for consistency.
*   Ensure all columns are stored using appropriate data types.

```python
df.isna().sum()
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
logger.info('filled the null values in salary column with mean')
df
```

```python
df.isna().sum()
```

```python
df.drop_duplicates(subset='Employee_ID')
```

```python
logger.info('dropped the duplicate values')
```

```python
df.rename(columns={
    'Employee_ID': 'employee_id',
    'Name': 'name',
    'Salary': 'salary',
    'Age': 'age',
    'Joining_Date': 'joining_date',
    'Resigned': 'resigned',
    'Department': 'department'
}, inplace=True)
```

```python
logger.info('renamed column into lowercase')
```

```python
df.dtypes
```

```python
df['name']= df['name'].astype('str')
df['department']= df['department'].astype('str')

logger.info('Typecasted the columns datatype')
```

### Task 3: Data Manipulation
*   Filter out employees who are above the age of 25 or have a salary above 40,000, and are not resigned.
*   Create a new column `YearsInCompany` by calculating the number of years each employee has worked in the company (from `JoiningDate` to today).
*   Record and log the number of records before and after filtering.

```python
fil1 = ((df['age'] > 25) | (df['salary'] > 40000)) & (df['resigned'] == False)
df[fil1]
```

```python
logger.info('Rows after filtering 1 is: ' + str(df[fil1].shape))
```

```python
df['years_in_company'] = (pd.Timestamp.now() - df['joining_date']).dt.days / 365
df
```

```python
logger.info('Added column years_in_comppany via timetstamp - joining date')
```

### Task 4: Aggregation & Analysis
*   Compute department-level statistics including average salary and median age.
*   Apply a salary increment of 10% for employees belonging to a specific department.
*   Log analytical results and changes applied.

```python
df.groupby('department').agg({'salary': 'mean', 'age': 'median'})
```

```python
logger.info('Computed department-level statistics including average salary and median age.')
```

```python
df.loc[df['department'] == 'IT', 'salary'] = df['salary'] * 1.10
df
```

```python
logger.info('it depertment got 10 percent increment')
```
