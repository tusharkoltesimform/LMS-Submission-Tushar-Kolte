```python
import re

with open('data_log.txt', 'r') as file:
    log_data = file.read()
```

Extract All IP Addresses
Write a regex to extract all the IP addresses from the log. Print the unique IP addresses in sorted order.

```python
ip_addresses = re.findall(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', log_data)
unique_ips = sorted(list(set(ip_addresses)))
for ip in unique_ips:
    print(ip)
```

 Extract User Actions

For all log lines containing "INFO," extract: The username (e.g., john_doe). The action they performed (e.g., 'UPDATE' or 'created a new record'). Store the results in a dictionary where the key is the username and the value is a unique list of their actions.

```python
content_lines = log_data.split('\n')

selected_lines = []

for ind in content_lines:
    if re.search(r'INFO', ind):
        selected_lines.append(ind)

splitted_lines = []

for s in selected_lines:
    splitted_lines.append(re.split(' ', s))

for i in range(len(splitted_lines)):
    splitted_lines[i] = splitted_lines[i][4:]

user_dict = {}
for row in splitted_lines:
    name = row[0]
    action = " ".join(row[1:])

    if name not in user_dict:
        user_dict[name] = []

    user_dict[name].append(action)

for iter in user_dict:
    i = 1
    print(f'{iter} :- ')
    for ind in user_dict[iter]:
        print(f'  {i}.{ind}')
        i+=1
    print('\n')
```

Validate and Extract Email Addresses

Search for valid email addresses in the log file. Ensure the email address adheres to standard rules (e.g., contains @, valid domain, no special characters in the username). Output the extracted email addresses.

```python
email_addresses = re.findall(r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+', log_data)
unique_emails = sorted(list(email_addresses))
print("Extracted Email Addresses:")
for email in unique_emails:
    print(email)
```

Extract Phone Numbers

Find and extract all phone numbers in the log. Ensure the phone numbers: Follow the format XXX-XXX-XXXX. Contain exactly 10 digits. Return the extracted numbers in a set.

```python
phone_numbers = re.findall(r'\b\d{3}-\d{3}-\d{4}\b', log_data)
for phone_number in phone_numbers:
    print(phone_number)
```

Extract All URLs
Search for and extract all URLs from the log. Ensure that: The URL starts with http:// or https://. It includes a valid domain name. Return the extracted URLs in a set.

```python
urls = re.findall(r'https?://(?:[a-zA-Z0-9-]+\.)+[a-zA-Z]{2,6}(?:/[^\s]*)?', log_data)

print("Extracted URLs:")
for url in urls:
    print(url)
```

Classify Log Levels

Using regex, classify each log entry by its severity level (INFO, WARNING, ERROR, or CRITICAL). Count the number of entries in each category and display the results in a dictionary

```python
log_levels = {}
for line in log_data.splitlines():
    match = re.search(r'\b(INFO|WARNING|ERROR|CRITICAL)\b', line)
    if match:
        level = match.group(1)
        log_levels[level] = log_levels.get(level, 0) + 1

print("Log Level Classification:")
print(log_levels)
```

Extract Timestamps

Extract all the timestamps (e.g., [2025-01-07 14:32:10]) from the log. Store the timestamps in a list and display them in ascending order.

```python
extracted_stamp = re.findall(r'\[(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2})\]',log_data)

for i in extracted_stamp:
  print(i)
```

Mask Sensitive Information

Create a new version of the log where: IP addresses are masked as .... Phone numbers are masked as XXX-XXX-XXXX. Email addresses are masked as hidden@example.com. Save the modified log to a new file named masked_data_log.txt.

```python
masked_log_data = log_data

masked_log_data = re.sub(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', '***.***.***.***', masked_log_data)
masked_log_data = re.sub(r'\b\d{3}-\d{3}-\d{4}\b', 'XXX-XXX-XXXX', masked_log_data)
masked_log_data = re.sub(r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-zA-Z]+', 'hidden@example.com', masked_log_data)

with open('masked_data_log.txt', 'w') as file:
    file.write(masked_log_data)
```

Validate Error Codes

Identify and validate error codes from the log. Error codes should: Start with DB_ERR_ followed by a 4-digit number. Print all unique valid error codes in a list.

```python
error_codes = re.findall(r'DB_ERR_\d{4}', log_data)
unique_error_codes = list(set(error_codes))

for code in unique_error_codes:
    print(code)
```

Parse Log into Structured Format

Using regex, parse the log entries into a structured format (dictionary or JSON) with the following keys: timestamp: The timestamp of the log entry. level: The log level (INFO, WARNING, ERROR, CRITICAL). message: The rest of the log message. Write the parsed data to a file named parsed_log.json.

```python
import json

content_lines = log_data.split('\n')
content_lines = content_lines[:-3]

splitted_lines = []

for s in content_lines:
    splitted_lines.append(re.split(' ', s))

list_of_logs = []

for row in splitted_lines:
    temp_dict = {}

    timestamp = " ".join(row[:2])
    timestamp = timestamp[1:-1]
    temp_dict['timestamp'] = timestamp

    level = row[2]
    temp_dict['level'] = level

    message = " ".join(row[3:])
    temp_dict['message'] = message

    list_of_logs.append(temp_dict)

with open('parsed_log.json', 'w') as f:
    json.dump(list_of_logs, f, indent=4)
```
