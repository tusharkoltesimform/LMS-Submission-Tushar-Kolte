Task 5: Update the File
1. Write a Python program to update the age of 'Sophie' to 23 in the students.txt file.
2. Save the updated data back to the file.


```python
nametofind = 'Sophie'
newlist = []

with open('students.txt', 'r') as f:
    for i in f:
        words = i.strip().split(', ')
        
        if words[0] == nametofind:
            
            words[1] = '23'
        
        
        newlist.append(", ".join(words) + "\n")


with open('students.txt', 'w') as fe:
    fe.writelines(newlist)


with open('students.txt', 'r') as f1:
    for i in f1:
        print(i.strip())
```

Task 6: Work with Large Files
1. Create a file named numbers.txt and write numbers from 1 to 1000, each on a new 
line.
2. Write a Python program to:
 Read the file.
 Calculate and display the sum of all numbers in the file.

```python
nums=[]
for i in range(1,101):
    nums.append(str(i) + '\n')

print(nums)

with open ('numbers.txt', 'w') as f:
    f.writelines(nums)
```

```python
sum = 0
with open ('numbers.txt', 'r') as f:
    for i in f:
      
       sum = sum + int(i.strip()) 

print(sum)
```

Task 8: Word Count
1. Create a file named paragraph.txt and add a paragraph of your choice.
2. Write a Python program to:
 Count the total number of words in the file.
 Count the occurrences of a specific word entered by the user.

```python
with open("paragraph.txt", "r") as file:
    content = file.read().lower()

words = content.split()

print("Total number of words:", len(words))

search_word = input("Enter a word to search: ").lower()

print("Occurrences:", words.count(search_word))
```
