Maximum three login attempts allowed
On failure after three attempts display: "Access Denied. Please contact admin."

```python


max_try = 3
attempt = 0
while attempt < max_try:
    print('Enter your username')
    username = input()
    print('Enter your password')
    pwd = input()

    if username == 'trainer1' and pwd =='Train@123':
        print("Welcome Trainer 1")
        break
    elif username == 'trainer2' and pwd =='Learn@123':
        print("Welcome Trainer 2")
        break
    else:
        attempt = attempt + 1
        print("incorrect credentials")

if attempt == 3:
    print('Access Denied. Please contact admin.')
```

3. Performance Evaluation
Calculate total and average marks for each trainee.
Grade classification:
≥ 85 : Excellent
70–84 : Good
50–69 : Average
< 50 : Needs Improvement

```python
trainees = []

trainee_count = int(input('How many Trainees: '))

for i in range(trainee_count):

    name = input('Enter Name: ')

    while True:
        marks = int(input('Enter python marks: '))
        if 0 <= marks <= 100:
            python_marks = marks
            break
        else:
            print("Invalid marks! Enter between 0 and 100.")

    while True:
        marks = int(input('Enter dsa marks: '))
        if 0 <= marks <= 100:
            dsa_marks = marks
            break
        else:
            print("Invalid marks! Enter between 0 and 100.")

    while True:
        marks = int(input('Enter control_flow marks: '))
        if 0 <= marks <= 100:
            control_flow_marks = marks
            break
        else:
            print("Invalid marks! Enter between 0 and 100.")

    total_marks = (python_marks + dsa_marks + control_flow_marks),
    avg = (python_marks + dsa_marks + control_flow_marks) / 3 

    if avg >= 85:
        grade = 'Excellent'
    elif  70 <= avg <=84 :
        grade = 'Good'
    elif 50 <= avg <= 69:
        grade = 'Average'
    else:
        grade = 'Needs Improvement'

    trainees.append({
        'name': name,
        'python_marks': python_marks,
        'dsa_marks': dsa_marks,
        'control_flow_marks': control_flow_marks,
        'total_marks': total_marks,
        'avg': avg,
        'grade': grade
    })

print(trainees)
```

4. Analytics & Reporting
Generate the following analytics:
Highest and lowest average scorer (without max/min)
Count of trainees per grade
Trainees who failed any subject (marks < 40)

```python
highest = trainees[0]
lowest = trainees[0]

for t in trainees:
    if t["avg"] > highest["avg"]:
        highest = t

    if t["avg"] < lowest["avg"]:
        lowest = t

print('Highest is', highest['name'], 'Lowest is', lowest['name'])
```

```python
grade_count = {
    "Excellent": 0,
    "Good": 0,
    "Average": 0,
    "Needs Improvement": 0
}

for t in trainees:
    grade_count[t["grade"]] += 1


print("Grade Distribution:")

for grade in grade_count:
    print(grade, ":", grade_count[grade])



failed_trainees = []

for t in trainees:
    if t["python_marks"] < 40 or t["dsa_marks"] < 40 or t["control_flow_marks"] < 40:
        failed_trainees.append(t["name"])


print("Trainees Failed")

if len(failed_trainees) == 0:
    print("None")
else:
    for name in failed_trainees:
        print(name)
```
